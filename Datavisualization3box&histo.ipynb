```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data",header=None,names=["sepal_length", "sepal_width", "petal_length", "petal_width", "class"])
```

```python
df.describe()
```

```python
print('Features of Iris Dataset:')
print('Sepal Length :',df['sepal_length'].dtype)
print('Sepal Width  :',df['sepal_width'].dtype)
print('Petal Length :',df['petal_length'].dtype)
print('Petal Width  :',df['petal_width'].dtype)
print('Species      :',df['class'].dtype)
```

```python
df.hist(figsize=(10, 8), edgecolor='black')
plt.suptitle("Histograms of Iris Dataset Features")
plt.show()
```

```python
plt.figure(figsize=(10, 6))
for i, column in enumerate(df.columns[:-1]):
    plt.subplot(2, 2, i+1)
    sns.boxplot(y=df[column])
    plt.title(f'Boxplot of {column}')
plt.tight_layout()
plt.show()
```

```python
4. Compare Distributions and Identify Outliers
From the histograms, we observe:

petal_length and petal_width are more spread and show clear clustering, potentially useful for classification.

sepal_width is more symmetrically distributed than others.

From the boxplots:

sepal_width shows some outliers (points falling outside whiskers).

The other features have a more compact range with minimal or no clear outliers.
```

```python
sns.histplot(x = df['sepal_length'], kde=True)
sns.histplot(y = df['sepal_width'], kde=True)
sns.histplot(x = df['petal_length'], kde=True)
sns.histplot(y = df['petal_width'], kde=True)
```
