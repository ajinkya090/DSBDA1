```python
df = pd.read_csv('iris.csv')
```

```python
species_list = df['Species'].unique()

# Loop through each species and show statistics
for species in species_list:
    print(f"Statistics for {species}:\n")
    
    # Filter rows for this species
    species_data = df[df['Species'] == species]
    
    # Describe gives count, mean, std, min, 25%, 50%, 75%, max
    print(species_data.describe())
    print("\n" + "-"*60 + "\n")
```

```python
def remove_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return df[(df[column] >= lower_bound) & (df[column] <= upper_bound)]

# Apply to each numerical column
for col in df.columns[:-1]:  # exclude target
    df = remove_outliers_iqr(df, col)

print("Data shape after removing outliers:", df.shape)
```

```python
sns.boxplot(x='SepalWidthCm',data=df) 
```
