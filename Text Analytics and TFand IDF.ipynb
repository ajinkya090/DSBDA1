```python
import nltk
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('stopwords')
from nltk import sent_tokenize
from nltk import word_tokenize
from nltk.corpus import stopwords
from nltk import pos_tag
```

```python
document='Real madrid is set to win the UCL for the season . Benzema might win Balon dor . Salah might be the runner up'
```

```python
tokens_sents = nltk.sent_tokenize(document)
print(tokens_sents)
```

```python
tokens_words = nltk.word_tokenize(document)
print(tokens_words)
```

```python
from nltk.stem import PorterStemmer
from nltk.stem.snowball import SnowballStemmer
from nltk.stem import LancasterStemmer
```

```python
stem=[]
for i in tokens_words:
  ps = PorterStemmer()
  stem_word= ps.stem(i)
  stem.append(stem_word)
print(stem)
```

```python
stemmer = PorterStemmer()
stemmed = [stemmer.stem(word) for word in stem]
print("Stemmed Tokens:", stemmed)
```

```python
from nltk.stem import WordNetLemmatizer 
lemmatizer = WordNetLemmatizer()
```

```python
lemmatized_output = ' '.join([lemmatizer.lemmatize(w) for w in stem])
print(lemmatized_output)
```

```python
print("Parts of Speech: ",nltk.pos_tag(stem))
```
